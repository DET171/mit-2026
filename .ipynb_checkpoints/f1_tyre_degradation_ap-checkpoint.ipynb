# F1 Tyre Degradation AP Model

We model tyre degradation with an arithmetic progression (AP):

$$T_n = T_0 + (n-1)d$$

- $T_n$: lap time on tyre lap $n$
- $T_0$: baseline lap time for that compound
- $d$: degradation rate (sec/lap)

This notebook is split into:

- **Part A**: extract degradation rates from FastF1 and store in a dictionary
- **Part B**: compute the most optimal pit strategy using those rates

```python
# If needed, run once:
# !pip install fastf1 pandas numpy matplotlib

import fastf1
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from itertools import product

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', 50)
```

```python
fastf1.Cache.enable_cache('./.fastf1_cache')
```

## Part A) Extract tyre degradation rates and pit-stop loss

```python
# ---------- Inputs ----------
YEAR = 2025
GRAND_PRIX = 'Abu Dhabi'
SESSION = 'R'
DRIVER = 'VER'

COMPOUNDS_TO_ANALYZE = ['SOFT', 'MEDIUM', 'HARD']
MIN_DEG_RATE_SEC_PER_LAP = 0.005
```

```python
session = fastf1.get_session(YEAR, GRAND_PRIX, SESSION)
session.load()

all_laps = session.laps.pick_drivers(DRIVER).copy()
display(all_laps[['Driver', 'LapNumber', 'Stint', 'Compound', 'LapTime', 'PitInTime', 'PitOutTime']].head())
```

```python
def prepare_nonpit_laps(laps_df):
    laps = laps_df.copy()
    laps = laps[laps['LapTime'].notna()].copy()
    laps['CompoundU'] = laps['Compound'].str.upper()
    laps['LapTimeSec'] = laps['LapTime'].dt.total_seconds()
    laps = laps[laps['PitInTime'].isna() & laps['PitOutTime'].isna()].copy()
    laps = laps.pick_quicklaps(threshold=1.07)
    laps = laps.sort_values(['Stint', 'LapNumber']).copy()
    laps['TyreLap'] = laps.groupby('Stint').cumcount() + 1
    laps['LapDeltaSec'] = laps.groupby('Stint')['LapTimeSec'].diff()
    return laps


def estimate_fuel_track_gain(nonpit_laps, early_lap_cutoff=4):
    # Early stint lap-to-lap deltas have minimal tyre wear, so they approximate fuel+track gain.
    early = nonpit_laps[nonpit_laps['TyreLap'] <= early_lap_cutoff].copy()
    deltas = early['LapDeltaSec'].dropna()
    if len(deltas) < 4:
        raise ValueError('Not enough early-lap deltas to estimate fuel/track gain.')
    return float(np.median(deltas))


def estimate_deg_rates(nonpit_laps, compounds, fuel_track_gain, min_deg=0.005):
    raw_deg = {}
    used_deg = {}
    base_lap = {}

    for c in compounds:
        dfc = nonpit_laps[nonpit_laps['CompoundU'] == c].copy()
        if len(dfc) < 6:
            continue

        # Deg estimate from median lap-to-lap increase minus global fuel/track gain
        deltas = dfc['LapDeltaSec'].dropna()
        if len(deltas) < 4:
            continue

        q1, q3 = np.quantile(deltas, [0.25, 0.75])
        iqr = q3 - q1
        low = q1 - 1.5 * iqr
        high = q3 + 1.5 * iqr
        deltas = deltas[(deltas >= low) & (deltas <= high)]
        if len(deltas) < 3:
            continue

        d_raw = float(np.median(deltas) - fuel_track_gain)
        d_used = float(max(d_raw, min_deg))

        # Base lap for AP in Part B: median of first usable tyre lap in each stint
        first_laps = dfc[dfc['TyreLap'] == 1]['LapTimeSec']
        if first_laps.empty:
            first_laps = dfc['LapTimeSec']

        raw_deg[c] = d_raw
        used_deg[c] = d_used
        base_lap[c] = float(first_laps.median())

    return raw_deg, used_deg, base_lap


def estimate_pit_loss_from_local_baseline(driver_laps):
    laps = driver_laps.copy()
    laps = laps[laps['LapTime'].notna()].sort_values('LapNumber').copy()
    laps['LapTimeSec'] = laps['LapTime'].dt.total_seconds()

    losses = []
    details = []

    for row in laps[laps['PitInTime'].notna()].itertuples(index=False):
        in_lap = int(row.LapNumber)
        out_lap = in_lap + 1
        out_rows = laps[laps['LapNumber'] == out_lap]
        if out_rows.empty:
            continue
        out_row = out_rows.iloc[0]

        pre_window = laps[
            (laps['LapNumber'] >= in_lap - 4)
            & (laps['LapNumber'] <= in_lap - 1)
            & laps['PitInTime'].isna()
            & laps['PitOutTime'].isna()
        ]

        post_window = laps[
            (laps['LapNumber'] >= out_lap + 1)
            & (laps['LapNumber'] <= out_lap + 4)
            & laps['PitInTime'].isna()
            & laps['PitOutTime'].isna()
        ]

        if pre_window.empty or post_window.empty:
            continue

        expected_pair = float(pre_window['LapTimeSec'].median() + post_window['LapTimeSec'].median())
        actual_pair = float(row.LapTimeSec + out_row['LapTimeSec'])
        loss = actual_pair - expected_pair

        if loss > 0:
            losses.append(loss)
            details.append({
                'pit_in_lap': in_lap,
                'pit_out_lap': out_lap,
                'pit_loss_seconds': loss
            })

    if not losses:
        raise ValueError('Could not estimate pit loss from local baseline windows.')

    return float(np.median(losses)), pd.DataFrame(details)


nonpit_laps = prepare_nonpit_laps(all_laps)
nonpit_laps = nonpit_laps[nonpit_laps['CompoundU'].isin(COMPOUNDS_TO_ANALYZE)].copy()

if nonpit_laps.empty:
    raise ValueError('No clean non-pit laps after filtering. Try another race/driver.')

race_progress_gain_sec_per_lap = estimate_fuel_track_gain(nonpit_laps)
raw_slope_by_compound, deg_rate_by_compound, base_lap_by_compound = estimate_deg_rates(
    nonpit_laps,
    compounds=COMPOUNDS_TO_ANALYZE,
    fuel_track_gain=race_progress_gain_sec_per_lap,
    min_deg=MIN_DEG_RATE_SEC_PER_LAP
)

if not deg_rate_by_compound:
    raise ValueError('No compounds had enough clean data to estimate degradation rates.')

PIT_STOP_LOSS_SECONDS, pit_loss_events_df = estimate_pit_loss_from_local_baseline(all_laps)

clean_laps_by_compound = {
    c: nonpit_laps[nonpit_laps['CompoundU'] == c].copy()
    for c in deg_rate_by_compound.keys()
}

# Required deliverable dictionary:
# keys = compound, values = degradation rate; and pit stop loss
model_inputs = {
    'deg_rate_by_compound': deg_rate_by_compound,
    'pit_stop_loss_seconds': PIT_STOP_LOSS_SECONDS,
    'fuel_track_gain_sec_per_lap': race_progress_gain_sec_per_lap
}

print('model_inputs =')
print(model_inputs)
print(f'Estimated fuel+track gain per race lap from data: {race_progress_gain_sec_per_lap:+.4f} sec/lap')
print(f'Estimated pit-stop loss from data (median): {PIT_STOP_LOSS_SECONDS:.2f} s')
display(pit_loss_events_df)

summary_df = pd.DataFrame({
    'compound': list(raw_slope_by_compound.keys()),
    'raw_deg_from_lap_deltas_sec_per_lap': [raw_slope_by_compound[c] for c in raw_slope_by_compound],
    'strategy_deg_used_sec_per_lap': [deg_rate_by_compound[c] for c in raw_slope_by_compound],
    'T0_for_partB_sec': [base_lap_by_compound[c] for c in raw_slope_by_compound]
}).sort_values('compound')
display(summary_df)
```

```python
plt.figure(figsize=(8, 4))
pd.Series(deg_rate_by_compound).sort_index().plot(kind='bar')
plt.ylabel('Degradation rate d (sec/lap)')
plt.title(f'Estimated tyre degradation rates from lap deltas ({DRIVER}, {YEAR} {GRAND_PRIX})')
plt.show()

plt.figure(figsize=(8, 4))
nonpit_laps['LapDeltaSec'].dropna().hist(bins=25)
plt.axvline(race_progress_gain_sec_per_lap, color='red', linestyle='--', label='fuel+track gain estimate')
plt.xlabel('Consecutive lap delta (sec)')
plt.ylabel('Count')
plt.legend()
plt.title('Lap-to-lap deltas used in Part A estimation')
plt.show()
```

## Part B) Calculate the most optimal pitting strategy

```python
TOTAL_LAPS = int(all_laps['LapNumber'].max())

# Search settings
MAX_STOPS = 3              # checks 0-stop to 3-stop strategies
MIN_STINT_LAPS = 8
REQUIRE_TWO_COMPOUNDS = True   # FIA dry-race rule style approximation

available_compounds = list(deg_rate_by_compound.keys())

def stint_time_ap(t0, d, n_laps):
    # Sum of AP: n/2 * (2*T0 + (n-1)d)
    return n_laps / 2 * (2 * t0 + (n_laps - 1) * d)


def total_strategy_time(stint_lengths, stint_compounds, base_lap_dict, deg_dict, pit_loss):
    if len(stint_lengths) != len(stint_compounds):
        raise ValueError('stint_lengths and stint_compounds must have same length')
    driving = 0.0
    for N, c in zip(stint_lengths, stint_compounds):
        driving += stint_time_ap(base_lap_dict[c], deg_dict[c], N)
    pit_stops = len(stint_lengths) - 1
    return driving + pit_stops * pit_loss


def generate_stint_partitions(total_laps, num_stints, min_stint):
    # Integer partitions into num_stints positive parts, each >= min_stint
    results = []

    def backtrack(remaining, k, current):
        if k == 1:
            if remaining >= min_stint:
                results.append(current + [remaining])
            return

        max_first = remaining - min_stint * (k - 1)
        for first in range(min_stint, max_first + 1):
            backtrack(remaining - first, k - 1, current + [first])

    backtrack(total_laps, num_stints, [])
    return results


candidates = []

for stops in range(0, MAX_STOPS + 1):
    stints = stops + 1
    length_options = generate_stint_partitions(TOTAL_LAPS, stints, MIN_STINT_LAPS)

    for lengths in length_options:
        for compounds_seq in product(available_compounds, repeat=stints):
            if REQUIRE_TWO_COMPOUNDS and len(set(compounds_seq)) < 2 and stints > 1:
                continue

            t = total_strategy_time(
                stint_lengths=lengths,
                stint_compounds=compounds_seq,
                base_lap_dict=base_lap_by_compound,
                deg_dict=deg_rate_by_compound,
                pit_loss=PIT_STOP_LOSS_SECONDS
            )

            candidates.append({
                'stops': stops,
                'stint_lengths': lengths,
                'stint_compounds': compounds_seq,
                'estimated_total_time_s': t
            })

if not candidates:
    raise ValueError('No valid strategies generated. Relax constraints.')

results_df = pd.DataFrame(candidates).sort_values('estimated_total_time_s').reset_index(drop=True)
best = results_df.iloc[0]

def format_strategy(lengths, compounds):
    return ' | '.join([f'{c}-{n}' for c, n in zip(compounds, lengths)])

results_df['strategy'] = [format_strategy(l, c) for l, c in zip(results_df['stint_lengths'], results_df['stint_compounds'])]

print(f'Total race laps used: {TOTAL_LAPS}')
print(f'Best strategy: {best["stint_compounds"]} with stint lengths {best["stint_lengths"]}')
print(f'Estimated total time: {best["estimated_total_time_s"]:.2f} s')

display(results_df[['stops', 'strategy', 'estimated_total_time_s']].head(10))
```

### Notes

- This is a simplified strategy optimizer using AP tyre degradation.
- It does not include traffic, undercut/overcut effects, safety cars, or weather.
- For your project, this is a good mathematical baseline and can be discussed with limitations.
